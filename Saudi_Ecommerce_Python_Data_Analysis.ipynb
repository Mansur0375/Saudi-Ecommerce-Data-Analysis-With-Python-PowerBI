# **PYTHON CHART**

```python
# IMPORTING MATPLOT FOR CREATING A CHARTS
import matplotlib.pyplot as plt
```

**THIS BAR CHART IS ANALYZE SALES BY CATEGORY**

```python

Category_Sales = df.groupby('Category')['Sales Amount'].sum().sort_values(ascending=False)
Category_Sales.plot(kind='bar')
plt.title('Category Sales')
plt.xlabel('Product Category')
plt.ylabel('Sales Amount (SAR)')
plt.legend
plt.show()
```

**THIS BAR CHART IS ANALYZE SALES BY CITIES**

```python

City_Sales=df.groupby('City')['Sales Amount'].sum().sort_values(ascending=False)
City_Sales.plot(kind='bar')
plt.title('City Sales')
plt.xlabel('City')
plt.ylabel('Sales Amount (SAR)')
plt.legend
plt.show()
```

**THIS BAR CHART IS ANALYZE SALES BY ORDER STATUS**

```python

Status_Count = df['Order Status'].value_counts()
Status_Count.plot(kind='pie',autopct="%1.1f%%")
plt.title('Order Status Count')
plt.xlabel('Order Status')
plt.ylabel('Count')
plt.show()
```

# **CHARTS ARE STORING IN A VARIABLES**

```python
# SALES BY CATEOGRY
ax=Category_Sales.plot(kind='bar')
ax.bar_label(ax.containers[0],fmt="%.2f")
ax.margins(y=0.15)
plt.title("Total Sales by Category")
plt.xlabel("Product Category")
plt.ylabel("Sales Amount (SAR)")
plt.show()
```

```python
# SALES BY CITY
city_sales = df.groupby("City")["Sales Amount"].sum().sort_values(ascending=True)
ax = city_sales.plot(kind="barh")
ax.bar_label(ax.containers[0], fmt="%.2f")
ax.margins(y=0.15,x=0.35)
plt.title("Total Sales by City")
plt.xlabel("City")
plt.ylabel("Sales Amount (SAR)")
plt.show()
```

# **ADVANCED DATA ANLAYSIS**







**FIND THE BEST SELLING PRODUCT**

```python
product_sales = df.groupby('Product')['Sales Amount'].sum().sort_values(ascending=False)
product_sales.head(10).reset_index()
```

**FIND THE BEST PROFITABLE PRODUCT**

```python
product_profit = df.groupby('Product')['Profit'].sum().sort_values(ascending=False)
product_profit.head(10).reset_index()
```

**TOP 10 PRODUCT SALES CHART**

```python
top_products = product_sales.sort_values(ascending=True)
top_products.head(10)

ax = top_products.plot (kind='barh')

ax.bar_label(ax.containers[0],fmt='%0.2f')
ax.margins(x=0.20)
plt.title('Top 10 Selling Products')
plt.xlabel('Sales Amount (SAR)')
plt.ylabel('Product')
plt.show()
```

# SALES BY REGION

```python
sales_region = df.groupby('Region')['Sales Amount'].sum()
sales_region.sort_values(ascending=False).reset_index()
```

```python
ax = sales_region.sort_values(ascending=False).plot(kind="bar")

ax.bar_label(ax.containers[0], fmt="%.2f")
ax.margins(y=0.15)

plt.title("Total Sales by Region")
plt.xlabel("Region")
plt.ylabel("Sales Amount (SAR)")
plt.show()
```

**CACULATING AVERAGE ORDER VALUE**

```python
total_sales = df["Sales Amount"].sum()
total_orders = df["Order ID"].nunique()

average_order_value = total_sales / total_orders

print("Average Order Value:", round(average_order_value, 2), "SAR")
```

**CALCULATING PROFIT MARGIN VALUE**

```python
total_sales = df["Sales Amount"].sum()
total_profit = df["Profit"].sum()

profit_margin = (total_profit / total_sales) * 100

print("Profit Margin:", round(profit_margin, 2), "%")
```

**CALCULATING SALES BY PAYMENT METHOD**

```python
payment_sales = df.groupby("Payment Method")["Sales Amount"].sum()

payment_sales.sort_values(ascending=False).reset_index()
```

**VISUALIZE PAYMENT METHOD SALES**

```python
ax = payment_sales.sort_values(ascending=False).plot(kind="barh")

ax.bar_label(ax.containers[0], fmt="%.2f")
ax.margins(y=0.15)

plt.title("Sales by Payment Method")
plt.xlabel("Payment Method")
plt.ylabel("Sales Amount (SAR)")
plt.show()
```
